```python
import numpy as np
import pandas as pd
import os.path as op
from os import listdir

bids_folder = '/Volumes/mrenkeED/data/ds-stressrisk'
fmriprep_confounds_include = ['framewise_displacement']
ses = 1

subFolders = [f for f in listdir(bids_folder) if f[0:3] == 'sub']

```

```python
sub = subFolders[0]
fd = []
for run in range(1,7):
        fmriprep_confounds_file = op.join(bids_folder,'derivatives', 'fmriprep', sub, f'ses-{ses}', 'func', f'{sub}_ses-{ses}_task-risk_run-{run}_desc-confounds_timeseries.tsv')
        fmriprep_confounds = pd.read_table(fmriprep_confounds_file)[fmriprep_confounds_include]
        fd.append(fmriprep_confounds.mean())

np.mean(fd)
        
```

```python
ses=1
fd_sub_ses1 = []
subs = []
for sub in  subFolders:
    fd = []                    
    for run in range(1,7):
            fmriprep_confounds_file = op.join(bids_folder,'derivatives', 'fmriprep', sub, f'ses-{ses}', 'func', f'{sub}_ses-{ses}_task-risk_run-{run}_desc-confounds_timeseries.tsv')
            fmriprep_confounds = pd.read_table(fmriprep_confounds_file)[fmriprep_confounds_include] 
            fd.append(fmriprep_confounds.mean())
    fd_sub_ses1.append(np.mean(fd))
    subs.append(int(sub[4:]))

ses=2
fd_sub_ses2 = []
for sub in  subFolders:
    fd = []                    
    for run in range(1,7):
            fmriprep_confounds_file = op.join(bids_folder,'derivatives', 'fmriprep', sub, f'ses-{ses}', 'func', f'{sub}_ses-{ses}_task-risk_run-{run}_desc-confounds_timeseries.tsv')
            fmriprep_confounds = pd.read_table(fmriprep_confounds_file)[fmriprep_confounds_include] 
            fd.append(fmriprep_confounds.mean())
    fd_sub_ses2.append(np.mean(fd))





    
```

```python
df_motion_1 = pd.DataFrame(data={'subject':subs,'session':1, 'motion':fd_sub_ses1, }).set_index('subject')
df_motion_2 = pd.DataFrame(data={'subject':subs,'session':2, 'motion':fd_sub_ses2, }).set_index('subject')
df_motion = df_motion_1.append(df_motion_2).set_index('session',append=True).sort_index(0,level='session')
```
