```python
import pandas as pd
import os.path as op
import os
import numpy as np
import nibabel as nb
from ipyparallel import Client
```

```python
engines = Client()
```

```python
home_dir =  '/home/dnl/habitization/'
data_dir = home_dir + '/data/'
subs = home_dir + 'subjects.txt'
subs = list(np.loadtxt(subs,str))

# Important: set variable of interest here
TEST = 'LogSV' 

new_folder = home_dir + 'templates'+'/glms_' + TEST + '_step2/'

sessions = ['a','b']
runs = map(str,range(1,7))

ignore_subs = []#'HAB02', 'HAB03', 'HAB05', 'HAB08', 'HAB10', 'HAB11', 'HAB12', 'HAB13', 'HAB14', 'HAB16','HAB01', 'HAB18', 'HAB19', 'HAB20', 'HAB21', 'HAB22', 'HAB23', 'HAB24', 'HAB25']
#First analysis included subjects 'HAB02', 'HAB03', 'HAB05', 'HAB08', 'HAB10', 'HAB11', 'HAB12', 'HAB13', 'HAB14', 'HAB16'
#Second analysis included subjects 'HAB01', 'HAB07', 'HAB17', 'HAB18', 'HAB19', 'HAB20', 'HAB21', 'HAB22', 'HAB23', 'HAB24', 'HAB25'
subs = [x for x in subs if x not in ignore_subs]
print(subs)
```

```python
#Copy all brain images for registration
for sub in subs: #finds and replaces info in text file that writes analysis
    for sesh in sessions:
        for run in runs:
            
            src = op.join(home_dir,'data',sub,'func', sesh + run + '.feat','reg')
            dst = op.join(home_dir,'data',sub,'func' + TEST , sesh + run + '.feat','reg')
            
            
            os.symlink(src, dst)                      
```

```python
def make_new_dir(dir_name):
    if not op.exists(dir_name):
        os.mkdir(dir_name)
```

```python
for sub in subs: #finds and replaces info in text file that writes analysis
    for sesh in sessions:        

            template = op.join(home_dir,'templates','glm_1variable_step2.fsf')
            new_f = op.join(home_dir,'templates','glms_' + TEST + '_step2','_'.join([sub,sesh]) +'.fsf')
            
            make_new_dir(new_folder)
            
            with open(template,'r') as file:
                filedata = file.read()

            filedata = filedata.replace('$SUB',sub)
            filedata = filedata.replace('$SESSION',sesh)
            filedata = filedata.replace('$TEST',TEST)
            
            #next get number of time points
            #func = op.join(home_dir,'data',sub,'func', sesh + run + '.feat','filtered_func_data.nii.gz')
            #func = nb.load(func)
            #ntpts = func.shape[-1]
            #filedata = filedata.replace('$NUM_TIMEPOINTS',str(ntpts))

            
            with open(new_f,'w') as file:
                file.write(filedata)
```

```python
def feat(in_tuple):
    sub,sesh = in_tuple
    fsf = op.join(new_folder,'_'.join([sub,sesh]) +'.fsf')
    
    cmd = ['feat',fsf]
    cmd = ' '.join(cmd)
    os.system(cmd)
    return in_tuple
```

```python
in_tuples = []
for sub in subs:
    for sesh in sessions:
        in_tuples.append((sub,sesh))
```

```python
sub_engines = engines[0:54] #less if you want to use fewer
sub_engines.push(dict(new_folder = new_folder))
sub_engines.execute('import os.path as op')
with sub_engines.sync_imports():
    import os
    
output = sub_engines.map_sync(feat,in_tuples)
```
