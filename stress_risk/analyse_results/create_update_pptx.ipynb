```python
# base env
from pptx import Presentation
import os.path as op
from os import listdir

bids_folder = '/Volumes/mrenkeED/data/ds-stressrisk'
im_folder = '/Volumes/mrenkeED/data/ds-stressrisk/figures/spm_2ndlevels' #op.join()
model_folder = op.join(bids_folder,'derivatives', 'spm', f'EU-3_con_2ndlevels')
cons = [f[14:] for f in listdir(model_folder)]
```

```python
from pptx.util import Inches
prs = Presentation() #'/Users/mrenke/Desktop/StressRisk/findings_03.pptx') # '/Users/mrenke/Desktop/test.pptx'
prs.slide_width = 12192000 # got it from reading in an existing presi and running prs.slide_width
prs.slide_height = 6858000

for model in ['EU', 'NLC']:
    for version in range(2,4):
        slide = prs.slides.add_slide(prs.slide_layouts[5]) # 5 = ONLY_TITLE, 6 = BLANK 
        slide.shapes.title.text = f'{model}-{version}'

        # col 1
        con = 'v1_ch'
        slide.shapes.add_picture(op.join(im_folder, f'{model}-{version}_{con}.png'),left= Inches(1.4/2.54), top = Inches(5/2.54), height=Inches(1.2))
        con = 'v2_unch'
        slide.shapes.add_picture(op.join(im_folder, f'{model}-{version}_{con}.png'),left= Inches(1.4/2.54), top = Inches(9.5/2.54), height=Inches(1.2))
        con = 'v1_ch_dif'
        slide.shapes.add_picture(op.join(im_folder, f'{model}-{version}_{con}.png'),left= Inches(1.4/2.54), top = Inches(14/2.54), height=Inches(1.2))
        
        # col 2
        con = 'v1_unch'
        slide.shapes.add_picture(op.join(im_folder, f'{model}-{version}_{con}.png'),left= Inches(18/2.54), top = Inches(5/2.54), height=Inches(1.2))
        con = 'v2_ch'
        slide.shapes.add_picture(op.join(im_folder, f'{model}-{version}_{con}.png'),left= Inches(18/2.54), top = Inches(9.5/2.54), height=Inches(1.2))
        con = 'v2_ch_dif'
        slide.shapes.add_picture(op.join(im_folder, f'{model}-{version}_{con}.png'),left= Inches(18/2.54), top = Inches(14/2.54), height=Inches(1.2))

prs.save('/Users/mrenke/Desktop/test.pptx')
```
