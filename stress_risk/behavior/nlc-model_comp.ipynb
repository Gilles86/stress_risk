```python
from stress_risk.behavior.utils import get_data
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import os.path as op
import bambi
import arviz as az

bids_folder = '/Users/mrenke/data/ds-stressrisk'

target_folder = op.join(bids_folder, 'derivatives', 'cogmodels')
model_folder = op.join(bids_folder, 'derivatives', 'cogmodels')

df = get_data().set_index('group', append=True)
group_mapping = df.groupby(['subject', 'group']).size().reset_index(['group']).drop(0, axis=1)

```

```python
# for model-1,.....
# shifts (subwise ses2-ses1) in priors
n_model = 1
idata1 = model_1

name_prior = 'risky_prior' # 'safe_prior' 
prior_mu_post = idata1.posterior[f'{name_prior}_mu'].to_dataframe()
prior_mu_post = prior_mu_post.join(group_mapping).set_index('group', append=True)

prior_mu_ses1_group0 = prior_mu_post.xs('Intercept', 0, f'{name_prior}_mu_regressors')
prior_mu_ses2_group0 = prior_mu_post.xs('Intercept', 0, f'{name_prior}_mu_regressors') + prior_mu_post.xs('session', 0, f'{name_prior}_mu_regressors')
prior_mu_ses1_group1 = prior_mu_post.xs('Intercept', 0, f'{name_prior}_mu_regressors') + prior_mu_post.xs('group', 0, f'{name_prior}_mu_regressors')
prior_mu_ses2_group1 = prior_mu_post.xs('Intercept', 0, f'{name_prior}_mu_regressors') + prior_mu_post.xs('group', 0, f'{name_prior}_mu_regressors') + prior_mu_post.xs('session', 0,f'{name_prior}_mu_regressors') + prior_mu_post.xs('session:group', 0, f'{name_prior}_mu_regressors')

prior_mu_ses1_group0 = prior_mu_ses1_group0.xs(0, 0, 'group')
prior_mu_ses2_group0 = prior_mu_ses2_group0.xs(0, 0, 'group')

prior_mu_ses1_group1 = prior_mu_ses1_group1.xs(1, 0, 'group')
prior_mu_ses2_group1 = prior_mu_ses2_group1.xs(1, 0, 'group')

df_change_prior_gr0 = (prior_mu_ses2_group0 - prior_mu_ses1_group0).groupby(['subject']).mean()
df_change_prior_gr1 = (prior_mu_ses2_group1 - prior_mu_ses1_group1).groupby(['subject']).mean()
df_change_prior = pd.concat([df_change_prior_gr0, df_change_prior_gr1], axis=0)
#df_change_prior.to_csv(f'/Users/mrenke/data/ds-stressrisk/interim_sum_data/subwise_change_{name_prior}_model-{n_model}.csv')
```

```python
# look at standard deviation

df_change_prior_sd_gr0 = (prior_mu_ses2_group0 - prior_mu_ses1_group0).groupby(['subject']).std()
df_change_prior_sd_gr1 = (prior_mu_ses2_group1 - prior_mu_ses1_group1).groupby(['subject']).std()
df_change_prior_sd_risky = pd.concat([df_change_prior_sd_gr0, df_change_prior_sd_gr1], axis=0)
```

```python
df_change_prior_sd_gr0 = (prior_mu_ses2_group0 - prior_mu_ses1_group0).groupby(['subject']).std()
df_change_prior_sd_gr1 = (prior_mu_ses2_group1 - prior_mu_ses1_group1).groupby(['subject']).std()
df_change_prior_sd_safe = pd.concat([df_change_prior_sd_gr0, df_change_prior_sd_gr1], axis=0)
```

```python
plt.hist(df_change_prior_sd_safe, label='safe')
plt.hist(df_change_prior_sd_risky, label='risky')
plt.xlabel('sd of posterior estimate of shift in prior')
plt.ylabel('frequency')
plt.legend(loc='upper right')
plt.show()
```

```python
# for model-5
n_model = 5
idata5 = model_5

name_prior = 'prior'
prior_mu_post = idata5.posterior[f'{name_prior}_mu'].to_dataframe()
prior_mu_post = prior_mu_post.join(group_mapping).set_index('group', append=True)

# prior_mu_ses1_group0 =
prior_mu_ses2_group0 = prior_mu_post.xs('session', 0, f'{name_prior}_mu_regressors')
prior_mu_ses1_group1 = prior_mu_post.xs('group', 0, f'{name_prior}_mu_regressors')
prior_mu_ses2_group1 = prior_mu_post.xs('group', 0, f'{name_prior}_mu_regressors') + prior_mu_post.xs('session', 0,f'{name_prior}_mu_regressors') + prior_mu_post.xs('session:group', 0, f'{name_prior}_mu_regressors')

#prior_mu_ses1_group0 = prior_mu_ses1_group0.xs(0, 0, 'group')
prior_mu_ses2_group0 = prior_mu_ses2_group0.xs(0, 0, 'group')

prior_mu_ses1_group1 = prior_mu_ses1_group1.xs(1, 0, 'group')
prior_mu_ses2_group1 = prior_mu_ses2_group1.xs(1, 0, 'group')

df_change_safe_prior_gr0 = prior_mu_ses2_group0.groupby(['subject']).mean() #(prior_mu_ses2_group0 - prior_mu_ses1_group0)
df_change_safe_prior_gr1 = (prior_mu_ses2_group1 - prior_mu_ses1_group1).groupby(['subject']).mean()
df_change_safe_prior = pd.concat([df_change_safe_prior_gr0, df_change_safe_prior_gr1], axis=0)
df_change_safe_prior.to_csv(f'/Users/mrenke/data/ds-stressrisk/interim_sum_data/subwise_change_{name_prior}_model-{n_model}.csv')
```
