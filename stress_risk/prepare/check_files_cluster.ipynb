```python
import glob
import re
import pandas as pd
from os import listdir

bids_folder = '/shares/zne.uzh/mrenke/ds-stressrisk'
subject_list = [int(f[4:]) for f in listdir(bids_folder) if f[0:3] == 'sub' and len(f) == 6]
all_subjects_set = set(subject_list)
```

### decoding variants

```python
key = 'decoded_pdfs.volume.cv_voxel_selection.denoise.retroicor'
ses = 'ses-2'
mask = 'NPC1_R'

fns = glob.glob(f'{bids_folder}/derivatives/{key}/sub-*/func/*_{ses}_mask-{mask}_*pars.tsv')
reg = re.compile(f'.*/sub-(?P<subject>[0-9]+)_({ses}_mask-{mask}_space-T1w_pars.tsv)')
```

```python
key = 'decoded_pdfs.volume.svoxels.ses1-en_ses2-de.denoise.retroicor'
ses = 'ses1-en_ses2-de'
mask = 'NPC1_R'

fns = glob.glob(f'{bids_folder}/derivatives/{key}/sub-*/func/*_{ses}_mask-{mask}_*.tsv')
reg = re.compile(rf'.*/sub-(?P<subject>[0-9]+)_{ses}_mask-{mask}_nvoxels-\d+_space-T1w_pars\.tsv') # nvoxels in file name problem
# reg = re.compile(rf'.*/sub-(?P<subject>[0-9]+)_(?P<session>{ses})_mask-{mask}_nvoxels-(?P<nvoxels>\d+)_space-T1w_pars\.tsv')
```

```python
data = []
for fn in fns:
    try:
        data.append(reg.match(fn).groupdict())
        data[-1]['fn'] = fn
    except Exception as e:
        print(f'Problem with {fn}: {e}')

data = pd.DataFrame(data)
data['subject'] = [int(sub) for sub in data['subject']]
df = data.set_index(['subject'])
```

```python
df =df.sort_values(by='subject')
df
```

```python
df_subjects_set = set(df.index)
missing_subjects = all_subjects_set - df_subjects_set
missing_subjects =list(missing_subjects)
print(*missing_subjects, sep=',')

# sacct -S '2023-02-20' -u gdehol -o jobid%25,start,exitcode,nodelist%25,state
```
