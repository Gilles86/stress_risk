```python
# copy all event files so that I can change them in the openNeuro BIDS folder

# move physio/value_event files from source to target BIDS folder
source_bids_folder = '/mnt_econ_department/projects/2022/renkert_dehollander_aydogan_ruff_stressrisk/data/ds-stressrisk_openNeuro'
target_bids_folder = '/mnt_econ_department/projects/2022/renkert_dehollander_aydogan_ruff_stressrisk/data/ds-stressrisk_remaining'

key = 'events.tsv' # 'physio.log'
import os 
subList = [sub[4:] for sub in os.listdir(source_bids_folder) if sub.startswith('sub-')]

for sub in subList:
    for ses in [1,2]:
        try:
            source_folder = os.path.join(source_bids_folder, f'sub-{sub}', f'ses-{ses}', 'func')
            target_folder = os.path.join(target_bids_folder, f'sub-{sub}', f'ses-{ses}', 'func')
            os.makedirs(target_folder, exist_ok=True)

            for file in os.listdir(source_folder):
                if (key in file): # leave 'T1w_deface.nii' there!
                    source_file_path = os.path.join(source_folder, file)
                    target_file_path = os.path.join(target_folder, file)
                    os.system(f'cp {source_file_path} {target_file_path}')
                    print(f'Copied {source_file_path} to {target_file_path}')
        except FileNotFoundError:
            print(f'No {key} files for sub-{sub} ses-{ses}')
```

```python
# Rebuild event files with correct duration column
from pathlib import Path
import os.path as op 
backup_root = Path(target_bids_folder)
target_bids = Path(source_bids_folder)

# Loop over subjects and sessions
for sub in subList:
    for ses in [1, 2]:

        target_func_dir = op.join(target_bids, f"sub-{sub}", f"ses-{ses}", "func")
        backup_func_dir = op.join(backup_root, f"sub-{sub}", f"ses-{ses}", "func")
        if not Path(target_func_dir).exists():
            continue
        
        event_files = sorted(Path(backup_func_dir).glob("*task-risk_run-*events.tsv"))
        event_files = [f for f in event_files if "value" not in f.name]


        for backup_file in event_files:
            # Load original events from backup (using a permissive encoding)
            df = pd.read_csv(backup_file, sep="\t", encoding="latin1")

            # Add duration column
            df["duration"] = 0.6
            df.loc[df["trial_type"] == "choice", "duration"] = 0.0

            # Force correct BIDS column order: onset, duration, everything else alphabetical
            cols = df.columns.tolist()
            base_cols = ["onset", "duration"]   
            other_cols = sorted([c for c in cols if c not in base_cols])
            df = df[base_cols + other_cols]

            # Write new file as UTF-8 (required by BIDS)
            dest_file = Path(target_func_dir) / backup_file.name
                        # delete old file to avoid server permission/write errors
            try:
                os.remove(dest_file)
            except Exception as e:
                print(f"    ⚠️ Could not delete old file: {e}")
                
            df.to_csv(dest_file, sep="\t", index=False, encoding="utf-8")
    print(f"Processed subject {sub} session {ses}")
```

```python
# Fixng IntendedFor fields in fieldmap JSON files

# Root of your dataset
dataset_root = Path(source_bids_folder) # target_bids = Path(source_bids_folder)

# Loop over subjects
for sub_dir in sorted(dataset_root.glob("sub-*")):
    sub_id = sub_dir.name  # e.g., sub-01
    
    # Loop over sessions
    for ses_dir in sorted(sub_dir.glob("ses-*")):
        ses_id = ses_dir.name  # e.g., ses-1

        fmap_dir = ses_dir / "fmap"
        if not fmap_dir.exists():
            continue

        # Loop over all JSON files in fmap
        for json_file in fmap_dir.glob("*.json"):
            try:
                with open(json_file, "r", encoding="utf-8") as f:
                    data = json.load(f)
            except json.JSONDecodeError:
                print(f"Skipping invalid JSON: {json_file}")
                continue
            
            if "IntendedFor" in data:
                intended_path = Path(data["IntendedFor"])
                # Keep session folder and filename, but replace sub-ID in filename
                filename = intended_path.name  # e.g., sub-01_ses-1_task-task_run-6_bold.nii
                parts = filename.split("_")
                if parts[0].startswith("sub-") and parts[0] != sub_id:
                    parts[0] = sub_id  # fix sub-ID
                    new_filename = "_".join(parts)
                    new_path = Path(ses_id) / intended_path.parent.name / new_filename
                    print(f"Fixing {json_file}: {data['IntendedFor']} -> {new_path}")
                    data["IntendedFor"] = str(new_path)

                    # Overwrite JSON
                    with open(json_file, "w", encoding="utf-8") as f:
                        json.dump(data, f, indent=4)
                            
            else:
                print(f"No IntendedFor in {json_file}, skipping")
```

```python
import json
from pathlib import Path

dataset_root = Path(source_bids_folder) 

for sub_dir in sorted(dataset_root.glob("sub-*")):
    sub_id = sub_dir.name  # e.g., sub-01

    for ses_dir in sorted(sub_dir.glob("ses-*")):
        ses_id = ses_dir.name  # e.g., ses-1

        fmap_dir = ses_dir / "fmap"
        if not fmap_dir.exists():
            continue

        for json_file in fmap_dir.glob("*.json"):
            with open(json_file, "r", encoding="utf-8") as f:
                data = json.load(f)

            intended_path = Path(data["IntendedFor"])
            filename = intended_path.name

            # Only fix files containing 'task-task'
            if "task-task" in filename:
                # Delete the old JSON to avoid overwrite issues
                json_file.unlink()
                
                # Fix sub-ID if necessary
                parts = filename.split("_")
                if parts[0].startswith("sub-") and parts[0] != sub_id:
                    parts[0] = sub_id

                # Replace task-task with task-risk
                parts = [p.replace("task-task", "task-risk") for p in parts]
                new_filename = "_".join(parts)

                # Preserve relative path (session folder + subfolder like func)
                new_path = Path(ses_id) / intended_path.parent.name / new_filename
                data["IntendedFor"] = str(new_path)

                # Write the corrected JSON
                with open(json_file, "w", encoding="utf-8") as f:
                    json.dump(data, f, indent=4)

                print(f"Fixed {json_file}: {filename} -> {new_filename}")
```
