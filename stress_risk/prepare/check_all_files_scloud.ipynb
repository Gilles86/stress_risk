```python
import glob
import re
import pandas as pd
```

```python
fns = glob.glob('/data/ds-stressrisk/sub-*/ses-*/func/*.tsv') # only sesssion 1 of behavior on sciencecloud
# fns = glob.glob('/scratch/mrenke/ds-stressrisk/sub-*/ses-*/func/*.tsv') 
```

```python
reg = re.compile('.*/sub-(?P<subject>[0-9]+)_ses-(?P<session>[0-9]+)_task-risk_run-(?P<run>[0-9]+)_(events.tsv)')
```

```python
data = []

for fn in fns:
    try:
        data.append(reg.match(fn).groupdict())
        data[-1]['fn'] = fn
    except Exception as e:
        print(f'Problem with {fn}: {e}')
```

```python
data = pd.DataFrame(data)
```

```python
data.set_index(['subject', 'session', 'run']).unstack('run')
```

```python
# fMRI files
fns = glob.glob('/data/ds-stressrisk/sub-*/ses-*/func/*.nii')

reg = re.compile('.*/sub-(?P<subject>[0-9]+)_ses-(?P<session>[0-9]+)_task-risk_run-(?P<run>[0-9]+)_(bold.nii)')
```
