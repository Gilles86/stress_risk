```python
import os
import os.path as op
import shutil

import glob
from tqdm.notebook import tqdm
import pandas as pd
import json
```

```python
bids_folder = '/Volumes/mrenkeED/data/ds-stressrisk'
#bids_folder = '/shares/zne.uzh/mrenke/ds-stressrisk'
subFolders = [f for f in os.listdir(bids_folder) if f[0:3] == 'sub']
```

```python
sub = 'sub-49'
ses = 'ses-1'
run = 1

json_files = glob.glob(op.join(bids_folder, sub, ses, 'func', '*.json'))
json_files
```

```python
for json_file in json_files:
    with open(json_file, 'r+') as f:
        data = json.load(f)
        print(data)
        modified = False
        for key in data:
            if data[key] == 'task':
                data[key] = 'risk'
                modified = True
                print(f'modified: {modified}')
        if modified:
            f.seek(0)
            json.dump(data, f)
            f.truncate()
```

```python
for sub in subFolders[40:]:
    for ses in ['ses-1', 'ses-2']:
        json_files = glob.glob(op.join(bids_folder, sub, ses, 'func', '*.json'))
        for json_file in json_files:
            with open(json_file, 'r+') as f:
                data = json.load(f)
                #print(data)
                modified = False
                for key in data:
                    if data[key] == 'task':
                        data[key] = 'risk'
                        modified = True
                        print(f'{sub},{ses}, modified: {modified}')
                if modified:
                    f.seek(0)
                    json.dump(data, f)
                    f.truncate()

       
```

```python
# finding and deleting files

file_pattern = os.path.join(bids_folder, 'sub-*', 'ses-*', 'func', '._sub*.json')
file_to_delete = glob.glob(file_pattern)

for file in file_to_delete:
    if os.path.exists(file):
        os.remove(file)
        print(f'Deleted file: {file}')
    else:
        print(f'The file does not exist: {file}')
```
