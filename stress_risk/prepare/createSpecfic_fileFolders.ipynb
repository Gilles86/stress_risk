```python
import os
import os.path as op
import shutil


bids_folder = '/Volumes/mrenkeED/data/ds-stressrisk'
subjects = [d for d in os.listdir(bids_folder) if os.path.isdir(os.path.join(bids_folder, d)) and 'sub-' in d]
```

```python
old_base = 'derivatives/fmriprep'
new_base = 'derivatives/fmriprep_onlyConfounds'

file_conditions = ['desc-confounds_timeseries.tsv']
```

```python

for sub in subjects:
    for ses in [1, 2]:
        old_dir = os.path.join(bids_folder, old_base, sub, f'ses-{ses}', 'func')
        new_dir = os.path.join(bids_folder, new_base, sub, f'ses-{ses}', 'func')
        
        # Check if the new directory path exists, if not, create it
        os.makedirs(new_dir, exist_ok=True)

        files = [f for f in os.listdir(old_dir) if os.path.isfile(os.path.join(old_dir, f))]
        for file in files:
            for specific_string in file_conditions:
                if specific_string in file:
                    shutil.copy2(os.path.join(old_dir, file), os.path.join(new_dir, file))
```
