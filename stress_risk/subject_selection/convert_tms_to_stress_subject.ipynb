```python
import os
import os.path as op
import shutil

import glob
from tqdm.notebook import tqdm
import pandas as pd
import json
```

```python
old_subjects = [1, 16, 50, 41, 52, 30, 65, 63]
new_subjects = [32, 53, 49, 50, 51, 54, 52, 61]

old_bids_folder = '/data/ds-tmsrisk'
new_bids_folder = '/data/ds-stressrisk'
```

```python
# Copy the event files for session 1

for new_subject, old_subject in tqdm(zip(new_subjects, old_subjects)):
    for subfolder in ['func', 'fmap', 'anat']:

        old_folder = op.join(old_bids_folder, f'sub-{old_subject:02d}', 'ses-1', subfolder)
        new_folder = op.join(new_bids_folder, f'sub-{new_subject:02d}', 'ses-1', subfolder)

        if not op.exists(new_folder):
            os.makedirs(new_folder)

        old_fns = glob.glob(op.join(old_folder, '*'))

        for old_fn in old_fns:
            old_path, old_fn = op.split(old_fn)

            new_fn = op.join(new_folder, old_fn.replace(f'sub-{old_subject:02d}', f'sub-{new_subject:02d}').replace('task-task', 'task-risk'))

            shutil.copy(op.join(old_path, old_fn), new_fn)

            if new_fn.endswith('json'):
                with open(new_fn, 'r') as f:
                    d = json.load(f)

                if 'task' in d:
                    d['task'] = 'risk'

                    with open(new_fn, 'w') as f:
                        json.dump(d, f)
```
